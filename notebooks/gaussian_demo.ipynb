# Interactive Gaussian Distribution

Use the sliders below to change the **mean (μ)** and **standard deviation (σ)** of a Gaussian distribution and watch the plot update live.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
```

```python
def plot_gaussian(mean=0.0, std=1.0):
    plt.close('all')
    x = np.linspace(mean - 4.5 * std, mean + 4.5 * std, 500)
    y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)

    fig, ax = plt.subplots(figsize=(9, 4))

    ax.plot(x, y, color='steelblue', linewidth=2.5)
    ax.fill_between(x, y, alpha=0.25, color='steelblue')
    ax.axvline(mean, color='tomato', linestyle='--', linewidth=1.8,
               label=f'μ = {mean:.2f}')
    ax.axvline(mean - std, color='gray', linestyle=':', linewidth=1.2)
    ax.axvline(mean + std, color='gray', linestyle=':', linewidth=1.2,
               label=f'σ = {std:.2f}')

    ax.set_xlabel('x', fontsize=12)
    ax.set_ylabel('Probability Density', fontsize=12)
    ax.set_title(f'Gaussian Distribution   N(μ={mean:.2f}, σ={std:.2f})', fontsize=13)
    ax.legend(fontsize=11)
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()


interact(
    plot_gaussian,
    mean=FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.1,
                     description='Mean (μ)',
                     style={'description_width': 'initial'},
                     layout={'width': '420px'}),
    std=FloatSlider(value=1.0, min=0.1, max=3.0, step=0.1,
                    description='Std Dev (σ)',
                    style={'description_width': 'initial'},
                    layout={'width': '420px'}),
)
```
